```python
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
llm = ChatGroq(model="llama-3.3-70b-versatile")
```

```python
class BlogState(TypedDict):
    title: str
    outline: str
    content: str
```

```python
def create_outline(state: BlogState) -> BlogState:
    
    # fetch titile
    title = state['title']
    
    #  call the lllm
    promt = f'Generate a detailed outline for a blog on the topic -{title}'
    outline = llm.invoke(promt).content
    
    # update state
    state['outline'] = outline
    
    return state
```

```python
def create_blog(state: BlogState) -> BlogState:
    
    # fetch titile
    title = state['title']
    outline = state['outline']
    
    #  call the lllm
    promt = f'Write a detailed content for a blog on the title -{title} using the following outline \n {outline}'
    content = llm.invoke(promt).content
    
    # update state
    state['content'] = content
    
    return state
```

```python
graph = StateGraph(BlogState)

# nodes
graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)

# edges

graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog',END)

workflow = graph.compile()
```

```python
initial_state = {'title':'Rise of AI in india'}

final_state = workflow.invoke(initial_state)
print(final_state)
```
